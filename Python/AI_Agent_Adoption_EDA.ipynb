# Handling Outliers

```python
import pandas as pd

# Select numerical columns
numeric_cols = df.select_dtypes(include="number").columns

# Store outlier counts
outlier_counts = {}

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_counts[col] = len(outliers)

# Display results
print(pd.Series(outlier_counts, name="Outlier_Count"))
```

```python
col = "Training_Hours"

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df[col] < lower_limit) |
    (df[col] > upper_limit)
]

print(outliers[["Record_ID", col]])
print("Total outliers:", len(outliers))
```

```python
import numpy as np

df["Log_Monthly_AI_Cost"] = np.log1p(
    df["Monthly_AI_Cost"]
)
```

```python
import pandas as pd

numeric_cols = df.select_dtypes(include="number").columns
results = []

for col in numeric_cols:
    values = df[col].dropna()

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((values < lower) | (values > upper)).sum()

    results.append({
        "Column": col,
        "Lower_Limit": lower,
        "Upper_Limit": upper,
        "Outlier_Count": count
    })

outlier_summary = pd.DataFrame(results)
print(outlier_summary)
```
