# Fill Numerical Columns with median

```python
df['age_group'] = df['age_group'].fillna(df['age_group'].mode()[0])

df['tasks_automated'] = df['tasks_automated'].fillna(df['tasks_automated'].median())

df['hours_saved_per_week'] = df['hours_saved_per_week'].fillna(df['hours_saved_per_week'].median())

df['productivity_change_percent'] = df['productivity_change_percent'].fillna(df['productivity_change_percent'].median())

df['training_hours'] = df['training_hours'].fillna(df['training_hours'].median())

df['monthly_ai_cost'] = df['monthly_ai_cost'].fillna(df['monthly_ai_cost'].median())

df['job_satisfaction'] = df['job_satisfaction'].fillna(df['job_satisfaction'].median())

df['trust_in_ai'] = df['trust_in_ai'].fillna(df['trust_in_ai'].median())

df['error_rate'] = df['error_rate'].fillna(df['error_rate'].median())
```

# Fill Categorical Columns with Mode

```python
df['ai_skill_level'] = df['ai_skill_level'].fillna(df['ai_skill_level'].mode()[0])
```

# Cleaning strings and normalizing missing values

```python
text_cols = df.select_dtypes(include=['object']).columns
na_representations = ["na", "n/a", "none", "null", "", "nan", "NaN", "NAT"]

for col in text_cols:
    # First replace text-based blanks with actual Pandas NaN while values are still objects
    df[col] = df[col].replace(na_representations, np.nan)
    
    # Then clean up the real text data (stripping whitespace and lowercasing)
    df[col] = df[col].astype(str).str.strip().str.lower()
    
    # Convert "nan" back to true np.nan just in case
    df[col] = df[col].replace("nan", np.nan)
```

# Imputing Missing Values

```python
text_cols = df.select_dtypes(include=['object']).columns
na_representations = ["na", "n/a", "none", "null", "", "nan", "NaN"]

for col in text_cols:
    df[col] = df[col].replace(na_representations, np.nan)
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# 4. Clean numerical columns with median imputation
num_cols = df.select_dtypes(include=['number']).columns
for col in num_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

# 5. Verify that all missing values are gone
print("Total remaining missing values:", df.isnull().sum().sum())
```

# Handling Outliers

```python
import pandas as pd

for col in df.select_dtypes(include='number').columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} outliers")
```
